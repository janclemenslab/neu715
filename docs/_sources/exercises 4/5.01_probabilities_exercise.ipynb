## Exercise 6: Visualize repeated spike trains with a raster plot

A **spike raster plot** shows when a neuron fires across repeated trials. Each horizontal row is one trial, and each short vertical tick marks one spike. The tick's horizontal position gives the spike time; its height does not represent spike size or duration. Dense groups of ticks indicate more spiking, and aligned ticks across rows reveal responses with similar timing. Unlike a histogram, a raster keeps the individual spike times and trials visible.

The supplied **synthetic data** contain responses from five different cells to the **same stimulus** (a pure tone). Each cell was recorded for 10 repeated presentations. The trials are grouped by cell: trials 1–10 are from cell 1, trials 11–20 from cell 2, and so on, giving 50 trials in total. The cells have different firing patterns, making their responses easy to compare.

Each trial lasts **1 second**, and the stimulus is on from **0.2 to 0.6 seconds**. Spike times are measured in **seconds from the start of each trial** (the preceding PSTH exercise uses milliseconds).

Run the loading cell below. As in assignment 1, there is one spike-time sequence and one matching label per trial. Here, the label identifies the cell:

- `data['spike_times'][i]` is a list of spike times for trial `i`.
- `data['cell_ids'][i]` identifies the cell recorded in that same trial.
- Python index `i = 0` corresponds to plotted trial number 1. Trials can contain different numbers of spikes, so keep their spike-time sequences separate.

Plot all 50 trials in a single raster, keeping their original block order.
