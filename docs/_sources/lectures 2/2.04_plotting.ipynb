## Plot a short voltage recording
Matplotlib supplies plotting functions. The import makes them available
under the short name `plt`; we use this setup line as supplied.

`plt.plot(x, y)` pairs values at matching indices: time on the horizontal
axis, voltage on the vertical axis.

````python
import matplotlib.pyplot as plt

times_ms = [0, 1, 2, 3, 4, 5]
voltage_mv = [-60, -62, 25, -58, -61, -60]
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
````

`xlabel` and `ylabel` describe the axes; `show` displays the figure.
The point at 2 ms has a voltage of 25 mV.

````python
other_times_ms = [0, 2, 5, 9]
other_voltage_mv = [-63, 22, -57, -64]
plt.plot(other_times_ms, other_voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
````

## Choose the visible range
`plt.xlim(1, 4)` shows times from 1 to 4 ms. It changes the view, not the data.
The limits must be set before `plt.show()`.

````python
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.xlim(1, 4)
plt.show()
````

## Mark a point
A second plot call can overlay markers. `"o"` draws circles as markers:

````python
plt.plot(times_ms, voltage_mv)
plt.plot(times_ms, voltage_mv, "o")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
````

## Put plots in separate panels
Subplots let us compare plots within one figure. Start a new figure with
`plt.figure()`, then select a panel with `plt.subplot(rows, columns, position)`.
The position starts at **1**, counting left to right across each row, then down.

`plt.subplot(1, 2, 1)` selects the left panel in a grid with one row and two columns;
`plt.subplot(1, 2, 2)` selects the right panel.
All following plotting calls, labels, titles, and limits apply to the selected panel.

Here, both panels show the same recording, but the right panel zooms in on the spike.


````python
plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.title("Full recording")
plt.ylim(-70, 40)

plt.subplot(1, 2, 2)
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.title("Spike close-up")
plt.xlim(1, 3)
plt.ylim(-70, 40)

plt.tight_layout()
plt.show()
````

`figsize=(8, 3)` sets the figure width and height in inches.
`plt.title` names the selected panel. `plt.tight_layout()` adjusts spacing
so labels fit; call it after setting up all panels and before `plt.show()`.
