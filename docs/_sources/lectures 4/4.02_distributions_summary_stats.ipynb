# Descriptive statistics

Measurements vary. Descriptive statistics help us summarize where a distribution is centered and how widely its values are spread.
The mean and standard deviation fully specify a *normal distribution*, but they cannot describe every distribution's shape.
We will look at the data first, then choose useful summaries.

## Measures of central tendency: Mean, median, mode

- **Mean:** the average, $\bar{x}=\frac{1}{N}\sum_{i=1}^{N}x_i$.
- **Median:** the middle value after sorting; for an even number of observations, average the two middle values.
- **Mode:** the most frequent discrete value, or a peak of a continuous probability density. There can be more than one mode.

For a normal population, mean, median, and mode coincide. For a skewed distribution they can differ substantially.
A finite sample only estimates these population features.

![Normal, exponential, and bimodal population densities with their means, medians, and modes.](fig/central_tendency.png)

**Discuss:** why is the exponential mean greater than its median? Is the mean a typical observation in the bimodal example?

For continuous measurements, a histogram's tallest bin gives only a bin-dependent estimate of the mode.
Calling `stats.mode` on mostly unique continuous values does not locate a useful density peak.

```python
scores = np.array([4, 5, 5, 6, 6, 7, 7, 8])
print('Mean:', np.mean(scores))
print('Median:', np.median(scores))
print('Modes:', 5, 6, 7)  # Each of these discrete values occurs twice.
```

## What does one extreme value change?

![A sample before and after adding a high value, with mean and median marked.](fig/extreme_value.png)

The mean uses every value's magnitude, so a large observation can pull it away from the main cluster.
The median depends on order and is often less sensitive to an extreme observation.

Here, adding one high value to just five observations moves the mean from **3.00 to 4.13**, while the median changes only from **3.00 to 3.10**.

## Measures of dispersion: Variance, standard deviation, and IQR

- **Sample variance:** $s^2=\frac{1}{N-1}\sum_{i=1}^{N}(x_i-\bar{x})^2$.
- **Sample standard deviation:** $s=\sqrt{s^2}$, in the same units as the observations.
- **Interquartile range:** $\mathrm{IQR}=Q_3-Q_1$, the width of the central 50% of the distribution.

For independent observations from the same population, dividing by $N-1$ makes the sample variance an unbiased estimator of population variance.
In NumPy, specify `ddof=1`; the default is division by $N$. 

![Normal and exponential distributions, comparing mean plus or minus SD with the interval from Q1 to Q3.](fig/dispersion.png)

For a normal population, about 68% of values fall within one SD of the mean and about 95% within two SDs.
Those percentages do **not** generally hold for skewed data. SD is still defined for such data, but median and IQR may summarize them more usefully.

```python
print('Sample variance:', np.var(scores, ddof=1))
print('Sample SD:', np.std(scores, ddof=1))
q1, q3 = np.percentile(scores, [25, 75])
print('Q1, Q3:', q1, q3)
print('IQR:', q3 - q1)
```

## Standard deviation versus standard error of the mean

**SD** describes the spread of individual observations. **SEM** estimates how much the sample mean would vary across repeated independent samples:

$$\mathrm{SEM}=\frac{s}{\sqrt{N}}.$$

**Discuss:** How SD and SEM change with the number of samples `N`?

![The same two sample means with SD, SEM, and 95 percent confidence interval error bars.](fig/error_bars.png)

Always label what error bars represent. A 95% confidence interval for a mean uses a multiplier; it is not mean ± SEM.
For independent normal observations, a t-based interval is $\bar{x}\pm t_{0.975,N-1}\,\mathrm{SEM}$.
Across repeated samples, 95% of intervals constructed this way would contain the fixed population mean under that model.

```python
sd = np.std(scores, ddof=1)
sem = sd / np.sqrt(len(scores))
print('SD:', sd, 'SEM:', sem)
```
