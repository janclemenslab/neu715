## 1. Create an environment

Python projects usually need additional packages: collections of code written for particular tasks. We have already used NumPy for arrays and Matplotlib for plotting. This week, we will also use Keras and PyTorch for machine learning.

These packages come in different versions, and some depend on particular versions of other packages. Installing or updating a package for one project can therefore cause problems for another project.

An environment is a separate installation of Python with its own collection of packages. For example, you could have one environment for this course and another for a research project. Each can use different Python and package versions.

Environments help us:
- Avoid conflicts: projects can use different package versions without interfering with each other.
- Try new tools: we can install machine-learning packages in a new environment while keeping an existing analysis setup working.
- Reproduce analyses: recording the Python and package versions helps us recreate a setup later or share it with someone else.
An environment contains the software needed to run your code. Your notebooks and data can stay in your usual project folder; they do not need to be stored inside the environment.

We will use conda to create and activate an environment called neu715-ai. Activating it selects which Python installation and packages commands in that terminal will use. The environment stays on your computer when you close the terminal; you activate it again next time.

The commands in the following blocks go into a **terminal**, one line at a time. On Windows, open the Miniconda Prompt or Anaconda Prompt. On macOS/Linux, open a terminal where conda is available.

```sh
conda create --name neu715-ai python=3.14 pip
conda activate neu715-ai
```

Confirm the installation with "yes" when prompted to do so. The first command creates the environment with Python 3.14. The second selects (activates) it for this terminal.

## 2. Install the packages

pip is Python’s package installer. It downloads and installs packages, usually from the Python Package Index ([PyPI](https://pypi.org)). It also installs additional packages needed by those packages, called dependencies.

Conda can install packages too. In this notebook, we use conda to create the environment and pip to install our tools inside it.

Keep `neu715-ai` active and run:

```sh
pip install keras torch numpy matplotlib jupyterlab notebook
```

- **Keras** is the interface we use to build and train the network.
- **PyTorch** does the calculations behind Keras.
- **NumPy and Matplotlib** handle arrays and plots, as in earlier weeks.
- **JupyterLab** opens our notebooks.

The downloads may take a few minutes.

## 3. Open JupyterLab

In the same terminal, move to the folder containing your downloaded notebooks. Replace the example path with your folder's path; keep the quotes if the path contains spaces.
