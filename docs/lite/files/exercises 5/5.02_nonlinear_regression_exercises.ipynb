## Exercise 1 - Fitting a quadratic function
The file `5.02_nonlinear_regression_exercises_1.npz` contains `x` and `y` values.

- Fit a quadratic function of the form  $y = a + b * (x + c)^2$ to describe the relationship between x and y. 
- Generate predictions (one for each x-value) and overlay them on the data. Is your curve a good fit to the data?
- Quantify the goodness of fit using the residual-based $R^2$ from the lecture and inspect the residuals.

```python
# your solution here
import numpy as np
import scipy.optimize
import matplotlib.pyplot as plt

def quadfun(x, a, b, c):
    y = a + b * (x + c)**2
    return y

d = np.load('5.02_nonlinear_regression_exercises_1.npz')
plt.plot(d['x'], d['y'], 'o')

initial_params = [-40, 10, 0]

popt, pcov = scipy.optimize.curve_fit(quadfun, d['x'], d['y'], initial_params)
popt

pred = quadfun(d['x'], popt[0], popt[1],  popt[2])
plt.plot(d['x'], pred)
r_squared = 1 - np.sum((d['y'] - pred)**2) / np.sum((d['y'] - np.mean(d['y']))**2)
print(f"Goodness of fit: R² = {r_squared:.3f}")
```

## Exercise 2 - Specifying initial parameter values
Below code attempts to fit an exponential function to the data. However, as you can see, the prediction does not match the data. Modify the initial parameter values used for fitting to make it work.

To do so, think about what how the different function parameters relate to specific properties of the data. Your function is $y = g * e^{-x/\tau}$. 
- The initial values are set by the gain parameter $g$. In other words: At $x=0$, the function evaluates to $y=g$. 
- $\tau$ corresponds to the "time" it takes until the function drops to 0.37 of the value at x=0. (The x-value at which y=0.37g) 

Using that knowledge you can estimate good starting values for $g$ and $\tau$ either visually or you can extract them directly from the data using simple heuristics.

```python
# Modify the initial parameter values to obtain a good fit.
def exp_fun(x, gain, tau):
    return gain * np.exp(-x / tau)

d = np.load('5.02_nonlinear_regression_exercises_2.npz')
x, y = d['x'], d['y']

initial_params = [1, 0.05]  # [gain, tau]: deliberately poor starting values
popt, pcov = scipy.optimize.curve_fit(exp_fun, x, y, p0=initial_params)
pred = exp_fun(x, *popt)

plt.plot(x, y, 'o', label='Data')
plt.plot(x, pred, label='Fitted curve')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
print(f"Fitted parameters: gain={popt[0]:.3f}, tau={popt[1]:.3f}")
```
