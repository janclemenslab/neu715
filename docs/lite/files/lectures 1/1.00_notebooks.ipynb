# Intro to notebooks

Learn how to run notebook cells, use Python as a calculator, and understand how cell order affects stored values.


## Cells

- **Code cells** contain Python code.
- **Markdown cells** contain explanations, headings, and questions.

Add a cell with the **+** button. Double-click a Markdown cell to edit it.
Run a cell with **Shift+Enter**; this also moves to the next cell.
**Ctrl+Enter** runs a cell and stays on it. The toolbar's run button works too.

An asterisk (`*`) beside a code cell means it is running. The value on the
last line of a code cell is displayed automatically.

## Use Python as a calculator

## Running order and notebook memory

You can run cells in any order, but later cells may depend on earlier
ones. Start at the top when opening a notebook. We will investigate
notebook memory after learning about variables and conditionals.
