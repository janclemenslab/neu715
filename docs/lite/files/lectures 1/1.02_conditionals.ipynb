# Conditionals

## Our question: does the animal meet the training rule?

An animal performs **20 trials** of a sound-detection task. On each trial,
a sound is either played or not played. The animal chooses one of two
response ports to report **"sound"** or **"no sound"**.

A response is **correct when it matches what happened**, including reporting
"no sound" when no sound was played.

For this exercise, we label an animal **well-trained** when it is correct on
**more than 75%** of the 20 trials: **more than 15 correct**, or an accuracy
**greater than 0.75**. Exactly 15 correct does **not** meet the rule.
This is the definition we use for this classroom task.


## Is a response correct?

`target` records what happened: `"sound"` or `"no sound"`.
`response` records the animal's report using the same labels.
A correct response matches the target.

`==` compares two values. It is different from `=`, which assigns a value.

````python
target = "sound"
response = "sound"
correct = response == target
print(correct)
````

A comparison produces `True` or `False`. This type is called `bool`, short
for Boolean. Predict the result if `response` becomes `"no sound"`.
Then change both labels to `"no sound"`. Why is that a correct response too?

````python
print(type(correct))
````

## Comparing numbers

| Operator | Meaning |
| --- | --- |
| `>` | greater than |
| `<` | less than |
| `==` | equal to |
| `!=` | not equal to |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

The training rule says **more than 15 correct**, so we use `>`, not `>=`.
Predict both results at the boundary.

````python
n_correct = 15
print(n_correct > 15)
print(n_correct >= 15)
````

**Try it:** use 14, 15, and 16. For which input do the two comparisons disagree?

Accuracy expresses the same rule as a fraction of the 20 trials.

````python
n_correct = 16
n_trials = 20
accuracy = n_correct / n_trials
well_trained = accuracy > 0.75
print(accuracy)
print(well_trained)
````

## Choose an action with if and else

We can choose an action depending on the outcome of a comparison using `if`/`else`. The basic logic of these statements is this: `if` EXPRESSION is True, then we do one thing, `else` we do something else:

The code below stores **1 for correct** and **0 for incorrect** trial outcomes.

````python
target = "sound"
response = "no sound"

if response == target:
    outcome = 1
    print("Correct")
else:
    outcome = 0
    print("Incorrect")

print(outcome)
````

Three parts of an `if`/`else` statement are new:

- **`if` and `else`:** The condition after `if` evaluates to `True` or `False`. If it is `True`, Python runs the code under `if`. Otherwise, it runs the code under `else`. Only one of these two blocks runs.
- **The colon (`:`):** A colon after the condition and after `else` tells Python that a block of code follows.
- **Indentation:** Indenting lines—usually with four spaces—shows which block they belong to. Lines that return to the original indentation are outside both blocks and run after the decision.

## Optional: combine comparisons

We can combine multiple boolean values using `and` and `or`.
`and` requires both comparisons to be True; `or` requires at least one.
For a completed 20-trial session, a correct-response count must be between
0 and 20, including both ends.

````python
n_correct = 14
print(n_correct >= 0 and n_correct <= 20)
print(n_correct < 0 or n_correct > 20)
````
