## Inspecting and clearing memory

The set of names Python currently knows is called its **namespace**.
In a Jupyter notebook, `%whos` displays the names and their values.

```python
%whos
```

Restarting the **kernel** clears this memory. The notebook's cells remain.
Old displayed output may remain too, so rerun the cells to check it.
