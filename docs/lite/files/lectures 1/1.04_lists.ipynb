## Slicing: access part of a list

A slice uses `list_name[start:stop]`. It includes the item at `start` and stops
**before** the item at `stop`.

**Worked example:** `letters[1:3]` uses indices 1 and 2, so it returns
`["b", "c"]`. Run the cell to see the result.

```python
worked_slice = letters[1:3]
print(worked_slice)
```

## List manipulation

Assign a new value to an indexed item to change it. This does not add a value to the list, it only changes an existing one.

## Optional: append versus extend

`append(value)` adds **one item** to a list. If the value is itself a list,
that whole list becomes one item. A list inside another list is a **nested list**.

`extend(other_list)` adds the **individual items** from another list.
Both methods change the list they are called on.

**Worked example:** compare the results and their lengths below. The same
input is added as one item with `append` and as separate items with `extend`.

```python
appended = ["a", "b"]
appended.append(["c", "d"])
print(appended)
print(len(appended))

extended = ["a", "b"]
extended.extend(["c", "d"])
print(extended)
print(len(extended))
```

**Predict:** what will each result contain, and how long will it be? Explain why.

```python
append_prediction = ["a"]
append_prediction.append(["b", "c", "d"])
print(append_prediction)
print(len(append_prediction))

extend_prediction = ["a"]
extend_prediction.extend(["b", "c", "d"])
print(extend_prediction)
print(len(extend_prediction))
```

## Optional: delete an item with del

## Optional: check membership with in

To ask whether a value is in a list, write `value in list_name`.
The result is `True` when a matching item exists and `False` otherwise.
`not in` asks the opposite question. These comparisons do not change the list
or report the matching item's index.

**Worked example:** check a letter that is present and one that is absent.

```python
membership_letters = ["a", "b", "c", "d", "e"]
print("a" in membership_letters)
print("z" in membership_letters)
```

**Predict:** what will the two comparisons below produce? Explain why.

```python
print("d" in membership_letters)
print("b" not in membership_letters)
```
