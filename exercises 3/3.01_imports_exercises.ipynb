## Exercise 3: save and test a different function

The code below uses the standard-library module `itertools`. Its `accumulate` function produces running totals; `list` collects those totals into a list. For example, counts `[2, 1, 0, 3]` have cumulative counts `[2, 3, 3, 6]`.

This function summarises trial counts rather than the outcome accuracy used in the lecture.

````python
import itertools

def cumulative_counts(counts):
    """Return the running total of trial counts."""
    return list(itertools.accumulate(counts))
````
