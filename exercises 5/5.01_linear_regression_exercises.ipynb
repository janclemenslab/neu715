## Exercise 1: Perform linear regression on two data sets
The file `5.01_linear_regression_exercises_1.npz` contains two datasets, `y1` and `y2` with common x-values in `x`.

For each dataset
- plot the data
- perform linear regression
- plot the regression line (predictions from the linear regression)
- inspect a plot of the residuals against x
- interpret the result - based on your results: are x and y1 (or y2) related in a meaningful way? are they linearly related?

```python
# your solution here
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats

d = np.load('5.01_linear_regression_exercises_1.npz')

plt.plot(d['x'], d['y1'], 'o')
plt.plot(d['x'], d['y2'], 'o')

res1 = scipy.stats.linregress(d['x'], d['y1'])
res2 = scipy.stats.linregress(d['x'], d['y2'])

plt.plot(d['x'], res1.slope * d['x']+ res1.intercept, 'b')
plt.plot(d['x'], res2.slope * d['x']+ res2.intercept, 'r')

res1.rvalue**2, res2.rvalue**2  # R² for each fitted line
```

## Exercise 2: Fit curves from multiple animals, compare params using statistics

The file `5.01_linear_regression_exercises_2.npz` contains two datasets, `y_ctrl` and `y_treat` with the x-values in `x`.
`y_ctrl` and `y_treat` are 100x10 matrices with control and treatment data from 10 animals per group:
- 100 rows correspond to the y values for each of the 100 x-values
- 10 columns correspond to data from 10 animals.

Your task is to:
- Perform linear regression on each of the 10 animals for the control and treatment data. You should get 10 slope and intercept parameters for the control and for the treatment data (20 slopes and 20 intercepts in total).
- Compare the 10 slope parameters for the control and treatment data using a plot.
- Test whether there exists a statistically significant difference between the regression parameters (slope or intercept) of the control and of the treatment animals.

For this exercise, treat control and treatment as independent groups of animals. Each animal contributes one slope to the group comparison; the 100 measurements per animal are not 100 independent animals.


### Exercise 2a: Perform linear regression on each animal

```python
# your solution here
d = np.load('5.01_linear_regression_exercises_2.npz')

nb_ctrl = d['y_ctrl'].shape[1]

ctrl_slope = []
ctrl_intercept = []
for animal in range(nb_ctrl):
    res = scipy.stats.linregress(d['x'], d['y_ctrl'][:, animal])
    ctrl_slope.append(res.slope)
    ctrl_intercept.append(res.intercept)    



nb_treat = d['y_treat'].shape[1]

treat_slope = []
treat_intercept = []
for animal in range(nb_treat):
    res = scipy.stats.linregress(d['x'], d['y_treat'][:, animal])
    treat_slope.append(res.slope)
    treat_intercept.append(res.intercept)    

treat_slope
```

### Exercise 2b: Visualize the regression parameters for the control and treatment animals

```python
# your solution here
plt.subplot(121)
plt.plot(np.zeros(len(ctrl_slope)), ctrl_slope, 'o')
plt.plot(np.ones(len(treat_slope)), treat_slope, 'o')
plt.subplot(122)
plt.plot(np.zeros(len(ctrl_slope)), ctrl_intercept, 'o')
plt.plot(np.ones(len(treat_slope)), treat_intercept, 'o')
```
