## Our question: does the animal meet the training rule?

An animal performs **20 trials** of a sound-detection task. On each trial,
a sound is either played or not played. The animal chooses one of two
response ports to report **"sound"** or **"no sound"**.

A response is **correct when it matches what happened**, including reporting
"no sound" when no sound was played.

For this exercise, we label an animal **well-trained** when it is correct on
**more than 75%** of the 20 trials: **more than 15 correct**, or an accuracy
**greater than 0.75**. Exactly 15 correct does **not** meet the rule.
This is the definition we use for this classroom task.

We can calculate directly with numbers. Here someone has already counted

### Give the values names

The above calculates the accurcay, but this is

1. not very general and 
2. not very readable. 

First, we directly tie the computation - calculating the accuracy - to the data values, if we want to compute the accuracy for another animal, we have to meddle with the code. This becomes more relevant if the computation becomes more complicated.

Second, it's unclear what `14 / 20` means without knowing the context of the code - these numbers could mean anything.

Variables allow you to express your program's computation in general terms, similar to how you use variables in math (`ax+b=c`).
With variables, you can apply the same computation do different values.

A variable is a name referring to a value. Create one with `name = value`.
`=` assigns the value on its right to the name on its left.
In math, the `=` is used to test for equality ($a = b$). In python, we use it for assigning values to variables. Double equal signs `==` test for equality in python.

Variables with informative names provide context - it's clear from looking at the variable names and the code what is being computed!

````python
n_correct = 14
n_trials = 20
accuracy = n_correct / n_trials
accuracy
````

`accuracy` is a **fraction**, not a count: 14 correct out of 20 gives 0.7.
Multiplying the fraction by 100 gives a percentage.

Python evaluates `n_correct / n_trials` using the values stored at that
moment, then stores the answer under `accuracy`.

**Predict:** change `n_correct` to 15 in the cell above and rerun the whole
cell. What should the new accuracy be? Then try 16 and restore 14.

````python
percent_correct = accuracy * 100
print(percent_correct)
````

### Comments and displayed output

it lets us show several results in one cell.

## Dealing with text

We also want to store the animal's name. Text needs single or double quotes.
Text values are called **strings**.

````python
animal_name = "Milo"
session_name = 'morning'
print(animal_name)
print(session_name)
````

**Predict the error:** why would `animal_name = Milo` fail in a fresh notebook?

Without quotes, Python treats `Milo` as another variable name. If that name
has not been assigned a value, Python raises a `NameError`.

When you code fails, python does not make your computer crash, but generates an error.
When this happens, read the final error line, locate the relevant cell, and
make one change before rerunning it. Error messages can be a bit cryptic sometimes.

### Determining the type of variables
Wrong variable types can be a source of errors in your code.

Below, an error is raised because one cannot add a variable of type string to a number in python.

````python
n_correct = 15
n_trials = '20'
n_correct / n_trials
````

It is therefore often useful to determine the type of a variable.

The `type` function returns the type of a variable:

## Types of values

- `int`: whole numbers, such as a count of 14 correct trials.
- `float`: decimal numbers, such as an accuracy of 0.7.
- `str`: text, such as an animal's name.

`type(...)` tells us a value's type. `print(...)` and `type(...)` are supplied
**functions**. 

Use a function by writing its name, followed by the input in parentheses:

```function_name(argument)```

Many functions also return a result, that can be assigned to a new variable: 

```result = function_name(argument)```


````python
n_correct = 14
accuracy = n_correct / 20
animal_name = "Milo"
print(type(n_correct))
print(type(accuracy))
print(type(animal_name))
````

### Numbers and text behave differently

Predict the two results. Why are they different?

````python
print(10 + 1)
print("10" + "1")
````

Adding numbers performs arithmetic. Adding strings joins the text.
Mixing them, as in `10 + "1"`, raises a `TypeError`.

### Converting a type

Use `int(...)` to turn whole-number text into an integer, `float(...)`
to turn numerical text into a decimal number, or `str(...)` to make text.

````python
count_as_text = "14"
n_correct = int(count_as_text)
print(n_correct / 20)
print(float("0.7"))
print("Animal " + str(3))
````

## Naming variables

Start a name with a letter or underscore. Use letters, digits, and
underscores after that. Names are case-sensitive: `n_correct` and
`N_correct` are different names.

- Useful: `animal_name`, `n_trials`, `n_correct`, `accuracy`.
- Hard to interpret: `a`, `b`, `x` for the same information.
- Invalid: `20_trials`, `number correct`, `n-correct`.

Some words, including `if` and `for`, are reserved by Python. Avoid using
function names such as `print` as your variable names too.
