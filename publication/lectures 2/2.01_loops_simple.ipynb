## Applying computations to list items using for-loops
In the above code, the only thing that changes is the index.

We can re-write this much more succinctly using a for-loop:

````python
letters = ["a", "b", "c", "d", "e"]
for index in [0, 1, 2, 3, 4]:
    print(letters[index])
print("Finished")
````

Here `index` takes the values 0, 1, etc in order. Note that indexing starts at 0 because the index of the first element in a list is 0, not 1. On each repetition, `letters[index]` retrieves the item at that position.

As for `if`/`else`, a colon ends the `for` statement and starts a code block, indentation marks which instructions should be repeated.
`"Finished"` is outside the block, so it is printed once after the loop.

### Manipulating list elements
Inside a for-loop, we can also manipulate list elements before printing them. For instance, convert weights from grams to kilograms:

````python
grams = [1000, 2000, 3000]

for index in [0, 1, 2]:
    kilograms = grams[index] / 1_000
    print(kilograms)
````

## Count correct responses
We can also update a result inside a for-loop, for instance by combining `if` and `for`. 

At each index, we compare the outcome to 1 to check whether it was correct. If so, we increase the `n_correct` count by 1. `n_correct = n_correct + 1` uses the old count and stores the increased value.

````python
practice_outcomes = [1, 0, 1, 1]
n_correct = 0  # initialize the count to 0
for index in range(len(practice_outcomes)):
    if practice_outcomes[index] == 1:  # is the outcome correct?
        n_correct = n_correct + 1  # increase to count by one
print(n_correct)
````

**Predict:** what count will this code produce? Compare this code to the code above and explain the difference.

````python
other_outcomes = [0, 1, 0, 1, 0]
count = 0
for index in range(len(other_outcomes)):
    if other_outcomes[index] == 0:
        count = count + 1
print(count)
````

## Build a list of results
We can also build a list of results using a for loop, by appending the result of calculations to a list. Can you predict what this code does?

````python
grams = [1000, 2000, 3000, 4000]
results = []

for gram in grams:
    kg = gram / 1_000
    results.append(kg)
    print(kg, results)
````

As before, we can combine this with `if`, to append only specific values to the results list:

````python
grams = [1000, 2000, 3000, 4000]
results = []

for gram in grams:
    kg = gram / 1_000
    if kg > 2:
        results.append(kg)
    print(kg, results)
````

## Optional: `While` loops
`For` loops are great if you have a pre-specified list of items and you want to do something with each item.Or if you have a computation and want to run it a fixed amount of times. However, sometimes you do not know beforehand how often you need to run a computation or you have an unspecified amount of items you need to work through. 

So-called `while` loops allow you to apply a computation as long as a condition is met. In that sense, they are a bit like a loopy version of `if`. While loops have the following form:
```python
while CONDITION:
    do something
```

For instance, we can use a while loop to count down to zero:

````python
count = 10

while count > 0:  # apply the indented code as long as the value of `count` is greater than 0
    count = count - 1
    print(count)
````

## Optional: An alternative way of using for loops
Above, we have used range objects to generate indices to access list elements in a for loop.

We access the elements of `indices`, one after the other, in the for loop. We then use the indices, to access the elements of `my_list`.

````python
letters = ["a", "b", "c"]
for index in range(len(letters)):
    print(letters[index])
````


However, we can access the list elements directly in a for loop:

````python


for letter in letters:
    print(letter)
````

## Optional: Generating more flexible ranges
So far, we have learned to use the range function to generate range-objects, by specifying the end: `range(stop)` will generate a range-object with indices starting at 0 and ending with stop-1.

However, there are two more ways to range: 
1. Specifying not only the end but also the start: `range(start, stop)`
1. Specifying the start, the end, and a step size: `range(start, stop, step)`

Specifying a range that way works a bit like specifying slice indices.

````python
print(list(range(0, 6, 2)))
````
