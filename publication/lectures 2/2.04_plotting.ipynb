## Plot a short voltage recording
Matplotlib supplies plotting functions. The import makes them available
under the short name `plt`; we use this setup line as supplied.

`plt.plot(x, y)` pairs values at matching indices: time on the horizontal
axis, voltage on the vertical axis.

```python
import matplotlib.pyplot as plt

times_ms = [0, 1, 2, 3, 4, 5]
voltage_mv = [-60, -62, 25, -58, -61, -60]
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
```

`xlabel` and `ylabel` describe the axes; `show` displays the figure.
The point at 2 ms has a voltage of 25 mV.

**Predict:** what point will appear at 5 ms in the new data below? Explain why.

```python
other_times_ms = [0, 2, 5, 9]
other_voltage_mv = [-63, 22, -57, -64]
plt.plot(other_times_ms, other_voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
```

## Choose the visible range
`plt.xlim(1, 4)` shows times from 1 to 4 ms. It changes the view, not the data.
The limits must be set before `plt.show()`.

```python
plt.plot(times_ms, voltage_mv)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.xlim(1, 4)
plt.show()
```

## Mark a point
A second plot call can overlay markers. `"o"` draws circles as markers:

```python
plt.plot(times_ms, voltage_mv)
plt.plot(times_ms, voltage_mv, "o")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane voltage (mV)")
plt.show()
```
